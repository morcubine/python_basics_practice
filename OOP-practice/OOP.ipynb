# Exercise 1 - Book Class

## Objective
Create a Python class named `Book` with the following attributes and methods.

### Attributes
1. `title` (str): The title of the book.
2. `author` (str): The name of the author.
3. `isbn` (str): The ISBN number of the book.
4. `available_copies` (int): The number of available copies of the book.

### Methods
1. `__init__(self, title, author, isbn, copies)`: Constructor to initialize the attributes.
2. `borrow_book(self)`: Decreases `available_copies` by 1 if the book is available. Otherwise, print "Book not available".
3. `return_book(self)`: Increases `available_copies` by 1.
4. `__str__(self)`: Returns a string in the following format: "Title: [title], Author: [author], ISBN: [isbn], Available Copies: [available_copies]".

## Sample Input
```python
# Creating an instance of Book
my_book = Book("1984", "George Orwell", "978-0451524935", 5)

# Borrowing a book
my_book.borrow_book()

# Printing the book details
print(my_book)

# Returning the book
my_book.return_book()

# Printing the book details again
print(my_book)
```

## Expected Output
```
Title: 1984, Author: George Orwell, ISBN: 978-0451524935, Available Copies: 4
Title: 1984, Author: George Orwell, ISBN: 978-0451524935, Available Copies: 5
```

````python
class Book:
    def __init__(self, title, author, isbn, copies): # parameters
        self.title = title
        self.author = author
        self.isbn = isbn
        self.copies = copies

    def borrow_book(self):
        if self.copies > 0:
            self.copies -= 1
        else:
            print("Book not available..")

    def return_book(self):
        self.copies += 1

    def __str__(self):
        return f"Title: {self.title}, Author: {self.author}, ISBN: {self.isbn}, Available Copies: {self.copies}"
````

# Exercise 2 - Rectangle Class

## Objective
Create a Python class named `Rectangle` with the following attributes and methods.

### Attributes
1. `length` (float): The length of the rectangle.
2. `width` (float): The width of the rectangle.

### Methods
1. `__init__(self, length, width)`: Constructor to initialize the attributes.
2. `area(self)`: Returns the area of the rectangle.
3. `perimeter(self)`: Returns the perimeter of the rectangle.
4. `__str__(self)`: Returns a string in the following format: "Rectangle with length [length] and width [width]".

## Sample Input
```python
# Creating an instance of Rectangle
my_rectangle = Rectangle(4.5, 2.0)

# Printing the rectangle details
print(my_rectangle)

# Printing the area

print("Area:", my_rectangle.area())

# Printing the perimeter
print("Perimeter:", my_rectangle.perimeter())
```

## Expected Output
```
Rectangle with length 4.5 and width 2.0
Area: 9.0
Perimeter: 13.0
```

````python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

    def __str__(self):
        return f'Rectangle with length {self.length} and width {self.width}'


r = Rectangle(4.5, 2.0)
print(r)
print(r.area())
print(r.perimeter())
````

# Exercise 3 - Student Class

## Objective
Create a Python class named `Student` with the following attributes and methods.

### Attributes
1. `name` (str): The name of the student.
2. `age` (int): The age of the student.
3. `grades` (list of int): A list of grades (integers).

### Methods
1. `__init__(self, name, age, grades)`: Constructor to initialize the attributes.
2. `add_grade(self, grade)`: Adds a new grade to the `grades` list.
3. `average_grade(self)`: Returns the average of the grades.
4. `__str__(self)`: Returns a string in the following format: "Student Name: [name], Age: [age], Average Grade: [average grade]".

## Sample Input
```python
# Creating an instance of Student
john = Student("John Doe", 20, [88, 92, 79])

# Adding a new grade
john.add_grade(85)

# Printing the student details
print(john)

# Printing the average grade
print("Average Grade:", john.average_grade())
```

## Expected Output
```
Student Name: John Doe, Age: 20, Average Grade: 86.0
Average Grade: 86.0
```

````python
class Student:
    def __init__(self, name, age, grades):
        self.name = name
        self.age = age
        self.grades = grades

    def add_grade(self, grade):
        self.grades.append(grade)

    def average_grade(self):
        return sum(self.grades) / len(self.grades)

    def __str__(self):
        return f"Student Name: {self.name}, Age: {self.age}, Average Grade: {self.average_grade()}"

john = Student("John Doe", 20, [88, 92, 79])
john.add_grade(85)
print(john)
print(f'Average Grade: {john.average_grade()}')
````

# Exercise 4 - BankAccount Class

## Objective
Create a Python class named `BankAccount` with the following attributes and methods.

### Attributes
1. `account_number` (str): The account number.
2. `account_holder` (str): The name of the account holder.
3. `balance` (float): The current balance of the account.

### Methods
1. `__init__(self, account_number, account_holder, balance)`: Constructor to initialize the attributes.
2. `deposit(self, amount)`: Adds the amount to the balance.
3. `withdraw(self, amount)`: Subtracts the amount from the balance if sufficient funds are available. Otherwise, print "Insufficient funds".
4. `__str__(self)`: Returns a string in the following format: "Account: [account_number], Holder: [account_holder], Balance: [balance]".

## Sample Input
```python
# Creating an instance of BankAccount
my_account = BankAccount("12345678", "John Doe", 1000.0)

# Making a deposit
my_account.deposit(500.0)

# Making a withdrawal
my_account.withdraw(200.0)

# Printing the account details
print(my_account)
```

## Expected Output
```
Account: 12345678, Holder: John Doe, Balance: 1300.0
```

````python
class BankAccount:

    def __init__(self, account_number, account_holder, balance):

        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        self.balance -= amount
        return self.balance

    def __str__(self):
        return f'Account: {self.account_number}, Holder: {self.account_holder}, Balance: {self.balance}'


my_account = BankAccount("12345678", 'John Doe', 1000.0)
my_account.deposit(500.0)
my_account.withdraw(200.0)

print(my_account)
````

# Exercise 5 - Timer Class

## Objective
Create a Python class named `Timer` with the following attributes and methods.

### Attributes
1. `hours` (int): The number of hours.
2. `minutes` (int): The number of minutes.
3. `seconds` (int): The number of seconds.

### Methods
1. `__init__(self, hours, minutes, seconds)`: Constructor to initialize the attributes.
2. `tick(self)`: Simulates a second passing, updating the `hours`, `minutes`, and `seconds` accordingly.
3. `__str__(self)`: Returns a string in the following format: "HH:MM:SS" (ensure two digits for each unit).

## Sample Input
```python
# Creating an instance of Timer
my_timer = Timer(1, 30, 5)

# Simulating the timer ticking
for _ in range(65):  # ticks for 65 seconds
    my_timer.tick()

# Printing the current time
print(my_timer)
```

## Expected Output
```
01:31:10
```

````python
class Timer:

    def __init__(self, hours, minutes, seconds):

        self.hours = hours
        self.minutes = minutes
        self.seconds = seconds

    def tick(self):
        self.seconds += 1
        if self.seconds > 60:
            self.minutes += 1
            self.seconds -= 60
            if self.minutes > 60:
                self.hours += 1
                self.minutes -= 60

    def __str__(self):
        return f'{self.hours:02d}:{self.minutes:02d}:{self.seconds:02d}'

my_timer = Timer(1, 30, 5)

for _ in range(65):
    my_timer.tick()

print(my_timer)
````

# Exercise 6 - ShoppingCart Class

## Objective
Create a Python class named `ShoppingCart` with the following attributes and methods.

### Attributes
1. `items` (dict): A dictionary where keys are item names (str) and values are their quantities (int).

### Methods
1. `__init__(self)`: Constructor to initialize the `items` dictionary as empty.
2. `add_item(self, item_name, quantity)`: Adds the specified quantity of the item to the cart. If the item already exists, increase the quantity.
3. `remove_item(self, item_name, quantity)`: Removes the specified quantity of the item from the cart. If the quantity exceeds the current amount, remove the item entirely.
4. `__str__(self)`: Returns a string representation of the cart in the following format: "Shopping Cart: [item1: quantity, item2: quantity, ...]".

## Sample Input
```python
# Creating an instance of ShoppingCart
cart = ShoppingCart()

# Adding items
cart.add_item("Apples", 5)
cart.add_item("Bananas", 3)
cart.add_item("Apples", 2)

# Removing an item
cart.remove_item("Bananas", 1)

# Printing the cart contents
print(cart)
```

## Expected Output
```
Shopping Cart: Apples: 7, Bananas: 2
```

````python
class ShoppingCart:

    def __init__(self, item_name, quantity):

        self.items = {}
        self.quantity = quantity


    def add_item(self, item_name, quantity):
        if item_name not in self.items:
            self.items[item_name] = quantity
        self.items[item_name] += quantity

    def remove_item(self, item_name, quantity):
        self.items[item_name] -= quantity

    def __str__(self):
        return f'Shopping Cart: {self.items}: {self.quantity}'

cart = ShoppingCart(_, _)

cart.add_item('Apples', 5)
cart.add_item('Bananas', 3)
cart.add_item('Apples', 2)

cart.remove_item('Bananas', 1)

print(cart)
````

# Exercise 7 - Library Class

## Objective
Create a Python class named `Library` with the following attributes and methods.

### Attributes
1. `books` (list): A list of books in the library. Each book is represented as a string (title of the book).

### Methods
1. `__init__(self)`: Constructor to initialize the `books` list as empty.
2. `add_book(self, book)`: Adds a book (a string) to the library.
3. `remove_book(self, book)`: Removes a book from the library. If the book is not in the library, print "Book not found".
4. `list_books(self)`: Prints the list of books in the library.
5. `__str__(self)`: Returns a string representation of the library in the following format: "Library: [book1, book2, ...]".

## Sample Input
```python
# Creating an instance of Library
my_library = Library()

# Adding books
my_library.add_book("1984")
my_library.add_book("To Kill a Mockingbird")

# Removing a book
my_library.remove_book("The Great Gatsby")

# Listing books
my_library.list_books()

# Printing the library contents
print(my_library)
```

## Expected Output
```
Book not found
1984
To Kill a Mockingbird
Library: 1984, To Kill a Mockingbird
```

````python
class Library:

    def __init__(self):
        self.book = []

    def add_book(self, book):
        self.book.append(book)

    def remove_book(self, book):
        if book in self.book:
            self.book -= 1
        else:
            print('Book Not Found')

    def list_books(self):
        return self.book

    def __str__(self): return f'Library: {self.book}'

my_library = Library()

my_library.add_book('1984')
my_library.add_book('To Kill a Mockingbird')

my_library.remove_book('The Great Gatsby')

print(my_library.list_books())

print(my_library)
````

# Exercise 8 - Playlist Class

## Objective
Create a Python class named `Playlist` with the following attributes and methods.

### Attributes
1. `songs` (list of dicts): A list where each element is a dictionary representing a song. Each dictionary has two keys: `title` (str) and `artist` (str).

### Methods
1. `__init__(self)`: Constructor to initialize the `songs` list as empty.
2. `add_song(self, title, artist)`: Adds a song to the playlist. The song is represented as a dictionary.
3. `remove_song(self, title)`: Removes a song from the playlist by title. If the song is not found, print "Song not found".
4. `get_total_songs(self)`: Returns the total number of songs in the playlist.
5. `__str__(self)`: Returns a string representation of the playlist in the following format: "Playlist: [Song Title - Artist, ...]".

## Sample Input
```python
# Creating an instance of Playlist
my_playlist = Playlist()

# Adding songs
my_playlist.add_song("Shape of You", "Ed Sheeran")
my_playlist.add_song("Believer", "Imagine Dragons")

# Removing a song
my_playlist.remove_song("Despacito")

# Printing the total number of songs
print("Total songs:", my_playlist.get_total_songs())

# Printing the playlist contents
print(my_playlist)
```

## Expected Output
```
Song not found
Total songs: 2
Playlist: Shape of You - Ed Sheeran, Believer - Imagine Dragons
```

````python
class Playlist:

    def __init__(self):
        self.songs = []

    def add_song(self, title, artist):
        self.songs.append({title, artist})

    def remove_song(self, title):
        if title not in self.songs:
            print('Song not found')
        else:
            self.songs.remove(title)

    def get_total_songs(self):
        return self.songs

    def __str__(self):
        return f'Playlist: {self.songs}'


my_playlist = Playlist()

my_playlist.add_song('Shape of You', 'Ed Sheeran')
my_playlist.add_song('Believer', 'Imagine Dragons')

my_playlist.remove_song('Despacito')

print(f'Total songs: {my_playlist.get_total_songs()}')

print(my_playlist)
````

# Exercise 9 - ContactBook Class

## Objective
Create a Python class named `ContactBook` with the following attributes and methods.

### Attributes
1. `contacts` (list of dicts): A list where each element is a dictionary representing a contact. Each dictionary has keys like `name` (str), `email` (str), and `phone` (str).

### Methods
1. `__init__(self)`: Constructor to initialize the `contacts` list as empty.
2. `add_contact(self, name, email, phone)`: Adds a contact to the contact book. The contact is represented as a dictionary.
3. `remove_contact(self, name)`: Removes a contact from the book by name. If the contact is not found, print "Contact not found".
4. `find_contact(self, name)`: Finds and returns the contact information for a given name. If the contact is not found, print "Contact not found".
5. `__str__(self)`: Returns a string representation of the contact book in the following format: "ContactBook: [Name: email, phone; ...]".

## Sample Input
```python
# Creating an instance of ContactBook
my_contacts = ContactBook()

# Adding contacts
my_contacts.add_contact("Alice", "alice@email.com", "123-456-7890")
my_contacts.add_contact("Bob", "bob@email.com", "098-765-4321")

# Removing a contact
my_contacts.remove_contact("Charlie")

# Finding a contact
print(my_contacts.find_contact("Alice"))

# Printing the contact book contents
print(my_contacts)
```

## Expected Output
```
Contact not found
{'name': 'Alice', 'email': 'alice@email.com', 'phone': '123-456-7890'}
ContactBook: Alice: alice@email.com, 123-456-7890; Bob: bob@email.com, 098-765-4321
```

````python
class ContactBook:

    def __init__(self):
        self.contacts = []

    def add_contact(self, name, email, phone):
        self.contacts.append({name, email, phone})

    def remove_contact(self, name):
        if name in self.contacts:
            self.contacts.remove(name)
        else:
            print('Contact not found')

    def find_contact(self, name):
        if name in self.contacts:
            print(self.contacts[name])
        else:
            print('Contact not found')

    def __str__(self):
        return f'ContactBook: {self.contacts}'


my_contacts = ContactBook()

my_contacts.add_contact('Alice', 'alice@email.com', '123-456-7890')
my_contacts.add_contact('Bob', 'bob@email.com', '098-765-4321')

my_contacts.remove_contact('Charlie')

print(my_contacts.find_contact('Alice'))

print(my_contacts)
````

# Exercise 10 - WeatherForecast Class

## Objective
Create a Python class named `WeatherForecast` with the following attributes and methods.

### Attributes
1. `forecasts` (dict): A dictionary where keys are dates (str, formatted as "YYYY-MM-DD") and values are weather descriptions (str).

### Methods
1. `__init__(self)`: Constructor to initialize the `forecasts` dictionary as empty.
2. `add_forecast(self, date, weather)`: Adds a weather forecast for a specific date.
3. `get_forecast(self, date)`: Returns the weather forecast for the given date. If no forecast is found for the date, print "No forecast available".
4. `__str__(self)`: Returns a string representation of the weather forecasts in the following format: "Weather Forecast: [YYYY-MM-DD: weather; ...]".

## Sample Input
```python
# Creating an instance of WeatherForecast
forecast = WeatherForecast()

# Adding forecasts
forecast.add_forecast("2023-11-20", "Sunny")
forecast.add_forecast("2023-11-21", "Partly cloudy")

# Getting a forecast
print(forecast.get_forecast("2023-11-20"))

# Printing all forecasts
print(forecast)
```

## Expected Output
```
Sunny
Weather Forecast: 2023-11-20: Sunny; 2023-11-21: Partly cloudy
```

````python
class WeatherForecast:

    def __init__(self):
        self.forecasts = {}


    def add_forecast(self, date, weather):  # YYYY-MM-DD: weather
        if date not in self.forecasts:
            self.forecasts[date] = weather


    def get_forecast(self, date):
        if date in self.forecasts:
            return self.forecasts[date]
        else:
            print('No forecast available')

    def __str__(self):
        return f'Weather Forecast: {self.forecasts}'


forecast = WeatherForecast()

forecast.add_forecast('2023-11-20', 'Sunny')
forecast.add_forecast('2023-11-21', 'Partly cloudy')

print(forecast.get_forecast('2023-11-20'))
print(forecast)
````
