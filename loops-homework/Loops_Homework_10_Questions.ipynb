## Question 6: Count the Increases

Count how many numbers are greater than the number immediately before them.

```python
numbers = [4, 7, 5, 9, 12, 8]
```

Here, `7`, `9`, and `12` are increases.

**Expected output:**
```text
3
```

````python
numbers = [4, 7, 5, 9, 12, 8]

# Write your solution here


count = 0

for i in range(len(numbers)):
    if numbers[i] < numbers[i-1]:
        count += 1

print(count)
````

## Question 7: Sum Values at Even Indexes

Find the sum of the numbers present at even indexes: `0, 2, 4...`

```python
numbers = [10, 5, 20, 8, 30, 6]
```

**Expected output:**
```text
60
```

Because `10 + 20 + 30 = 60`.

````python
numbers = [10, 5, 20, 8, 30, 6]

# Write your solution here

total = 0

for i in range(len(numbers)):
    if i % 2 == 0:
        total += numbers[i]

print(total)
````

## Question 8: Create a List of Word Lengths

## Question 12: Find the sum of digits.

```Example:

Input: 4821
```

````python
num = 4821

s = str(num)
total = 0

for n in s:
    total += int(n)
print(total)
````

## Question 14: Check whether a number is a palindrome.

```Example:

121 → Yes
123 → No
```

````python
# take input

n = 121
# n = 123

s = str(n)

if s == s[::-1]:
    print('Yes')
else:
    print('No')
````
