```python
# 10. sort a list of integers according to the sum of their digits.

nums = [3, 12, 25, 7, 40, 33, 51, 9, 28, 17, 49, 2, 31, 26, 5]
sum_digits = []



for num in nums:
    sum_num = 0
    for digit in str(num):
        sum_num += int(digit)
    # print(sum_num)
    sum_digits.append((num, sum_num))

sorted_sums = sorted(sum_digits, key=lambda x: x[1])

zero_i = [x[0] for x in sorted_sums]

print(zero_i)

```

```python
# 5. Palindrome Words
# Print all palindrome words from a sentence.
# Example: "madam likes level roads" → madam, level

sentence = "madam likes level roads and racecar tracks at noon"
words = sentence.split()
# print(words)
palin_words = []

for word in words:
    # print(word)
    reverse = word[::-1]
    # print(palindrome)
    if word == reverse:
        palin_words.append(word)

print(', '.join(palin_words))
```

```python
# 6. String Compression
# Compress consecutive repeating characters.
# Example: "aaabbccccdaa" → "a3b2c4d1a2"

s = "aaabbccccdaaa"
lst_a = []
lst_b = []
count = 1
new_s = ''

for i in range(1, len(s)):
    if s[i] == s[i-1]:
        count += 1
    else:
        lst_a.append(s[i-1])
        lst_b.append(count)
        count = 1


lst_a.append(s[-1])
lst_b.append(count)


for i in range(len(lst_a)):
    new_s += lst_a[i]
    new_s += str(lst_b[i])



# print(lst_a, lst_b)
print(new_s)

```

```python
# 7. Anagram Check
# Check whether two strings are anagrams without using sorted().
# Example: "listen" and "silent" → True

str1 = "listen"
str2 = "silent"
freq1 = {}
freq2 = {}

if len(str1) != len(str2):
    print(False)
else:
    for letter in str1:
        if letter not in freq1:
            freq1[letter] = 0
        freq1[letter] += 1

    for letter in str2:
        if letter not in freq2:
            freq2[letter] = 0
        freq2[letter] += 1

    if freq1 == freq2:
        print(True)
    else:
        print(False)
```

```python
# 8. Move Zeros to End
# Move all zeros to the end while preserving the order of other elements.
# Example: [0, 1, 0, 3, 12] → [1, 3, 12, 0, 0]

nums = [0, 1, 0, 3, 12]
new_nums = []
zeros = []


for num in nums:
    if num != 0:
        new_nums.append(num)
    else:
        zeros.append(num)

new_nums.extend(zeros)

print(new_nums)
```
